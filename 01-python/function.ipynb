# map

```python
def is_even(n):
    return n%2==0

num=[23,3,6,2,7,4,9]
even=list(filter(is_even,num))
double=list(map(is_even,even))
print(even)
print(double)
```

```python
num=[23,3,6,2,7,4,9]
even=list(filter(lambda x:x%2==0,num))
double=list(map(lambda n:n*2,even))
print(even)
print(double)
```

# python decorator 

```python
def div(a,b):
    print(a//b)

div(4,2)
```

```python
def div(a,b):
    print(a//b)

div(2,4)
```

```python
def div(a,b):
    if a<b:
        a,b=b,a
    print(a//b)

div(2,4)
```
