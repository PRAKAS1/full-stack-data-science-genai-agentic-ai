# variable length in arguments

```python
def sum(a,b):
    c=a+b
    print(c)

sum(5,6,7,8,9)    
```

```python
def sum(a,*b):
    c=a+b
    print(c)

sum(5,6,7,8,9)   
```

```python
def sum(a,*b):
    print(type(a))
    print(type(b))

sum(5,6,7,8,9)   
```

```python
def sum(a,*b):
    c=a
    for i in b:
        c=c+i
    print(c)    

sum(5,6,7,8,9)  
```

```python
def person(name,*data):
    print(name)
    print(data)

person('Alex',29,678678,'Sin')
```

```python
def person(name,*data):
    print(name)
    print(data)

person('Alex',age=29,add='patna')
```
